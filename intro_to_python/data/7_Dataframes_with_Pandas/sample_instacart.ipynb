```python
import pandas as pd
import random
```

```python
orders_raw = pd.read_csv('data/7_Dataframes_with_Pandas/orders.csv')
random.seed(123)
users = random.sample(list(orders_raw["user_id"]), k=int(len(set(orders_raw["user_id"]))*.025))
orders = orders_raw[orders_raw["user_id"].isin(users)]
order_products = pd.read_csv("data/7_Dataframes_with_Pandas/order_products__prior.csv")
order_products_new = order_products[order_products["order_id"].isin(orders["order_id"])]
len(set(order_products_new["order_id"])), len(set(order_products_new["order_id"]))
orders.drop("eval_set",inplace=True, axis=1)
orders.to_csv("data/7_Dataframes_with_Pandas/orders.csv", index=False)
order_products_new.to_csv("data/7_Dataframes_with_Pandas/orders_products.csv", index=False)
```
