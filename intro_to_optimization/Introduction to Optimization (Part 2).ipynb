## Tables-and-chairs revisited
Here is the tables-and-chairs example we saw previously, together with the code:

$$
\begin{align*}
\max_{x_1,x_2} \quad
& 20 x_1 + 30 x_2 
\\
\text{such that} \quad
& x_1 + 2 x_2 \leq 30 \\
& 10 x_1 + 5 x_2 \leq 120 \\
& x_1, x_2 \geq 0.
\end{align*}
$$

```julia
model = Model(Gurobi.Optimizer)
@variable(model, x[1:2] >= 0)
@objective(model, Max, 20x[1] + 30x[2])
# Named constraint
@constraint(model, labor, x[1] + 2x[2] <= 30)
@constraint(model, wood, 10x[1] + 5x[2] <= 120)
optimize!(model)
```

We saw last time how to query the optimal solution and optimal objective values:

```julia
println("Optimal objective value: ", objective_value(model))
println("Optimal solution: x1 = ", value(x[1]), ", x2 = ", value(x[2]))
```

What's new here is we can also directly query the dual solution from the model. There are various ways to do this:

```julia
println("Optimal dual solution for labor constraint: (using shadow_price()): ", shadow_price(labor))
println("Optimal dual solution for wood constraint: (using shadow_price()):  ", shadow_price(wood))
```

```julia
println("Optimal dual solution for labor constraint: (using dual()): ", dual(labor))
println("Optimal dual solution for wood constraint: (using dual()):  ", dual(wood))
```

Why are there two functions and why do they do different things? This is unique to the JuMP package; they have a `dual` function which works more generally but is independent of the objective sense (Max or Min), and a `shadow_price` function which works specific to linear programming. Read more here: https://jump.dev/JuMP.jl/stable/api/JuMP/#shadow_price

## Exercise 1

Here is the dual of the tables-and-chairs problem:

$$
\begin{align*}
\min_{p_1,p_2} \quad
& 30 p_1 + 120 p_2 
\\
\text{such that} \quad
& p_1 + 10 p_2 \geq 20
\\
& 2 p_1 + 5 p_2 \geq 30 
\\
& p_1, p_2 \geq 0
\end{align*}
$$

(a) Formulate and solve the dual problem, and verify that you get the same dual solution as before.

```julia
model_dual = Model(Gurobi.Optimizer)
@variable(model_dual, p[1:2] >= 0)
@objective(model_dual, Min, 30p[1] + 120p[2])
@constraint(model_dual, cost_table, p[1] + 10p[2] >= 20)
@constraint(model_dual, cost_chairs, 2p[1] + 5p[2] >= 30)
optimize!(model_dual)
```

```julia
p_sol = value.(p)
println("Optimal objective value of dual: ", objective_value(model_dual))
println("Optimal solution of dual: p1 = ", p_sol[1], ", p2 = ", p_sol[2])
```

(b) Plot the feasible region and optimal solution of the dual problem. 

```julia
xmax = 30
ymax = 15

Plots.plot(
    xlabel="p_1 (Price of labor)", 
    ylabel="p_2 (Price of wood)",  
    legend=:topright,
    aspect_ratio=:equal,
    xlims=(0, xmax),
    ylims=(0, ymax),
)
feasible_region = Shape(
    [ 
        (xmax, 0),
        (20 / 1, 0),
        Tuple([20, 30]' * inv([1 2; 10 5])),
        (0, 30 / 5),
        (0, ymax),
        (xmax, ymax),
    ]
)
Plots.plot!(
    feasible_region, 
    fillcolor=:gray,
    fillalpha=0.5,
    label=false,
)
Plots.plot!(
    [(x, (20-x)/10) for x in 0:0.1:xmax], 
    label="p1 + 10p2 = 20", 
    linewidth=3,
)
Plots.plot!(
    [(x, (30-2*x)/5) for x in 0:0.1:xmax], 
    label="2p1 + 5p2 = 30", 
    linewidth=3,
)
Plots.plot!(
    [(0, y) for y in 0:0.1:ymax], 
    label="p1 = 0", 
    linewidth=3,
)
Plots.plot!(
    [(x, 0) for x in 0:0.1:xmax], 
    label="p2 = 0", 
    linewidth=3,
)
Plots.scatter!(
    [p_sol[1]], 
    [p_sol[2]], 
    label="Optimal Solution", 
    color=:black, 
    markersize=5
)
Plots.annotate!(
    p_sol[1] + 4, 
    p_sol[2] + 1, 
    text(
        "($(string(round(p_sol[1], digits=3))), $(string(round(p_sol[2], digits=3))))",
        :black,
        8,
        :center,
    )
)
```

## Exercise 2: Sensitivity analysis

We saw the interpretation of the shadow price of labor and wood: each additional unit of labor and increases profit by $\frac{40}{3}$ and $\frac{2}{3}$ respectively. When does this hold until?

Hint: write the labor-wood LP as a function that takes two parameters: the amount of labor availability and wood availability, and returns the optimal primal solution and dual solution. Then use that function to test different values and form a hypothesis.

Here I've put the tables-and-wood linear program into its own function, which takes two parameters: the amount of labor availability and wood availability respectively. 

```julia
function solve_labor_wood_LP(
    labor_avail::Float64,
    wood_avail::Float64,
)
    model = Model(Gurobi.Optimizer)
    @variable(model, x[1:2] >= 0)
    @objective(model, Max, 20x[1] + 30x[2])
    @constraint(model, labor, x[1] + 2x[2] <= labor_avail)
    @constraint(model, wood, 10x[1] + 5x[2] <= wood_avail)
    optimize!(model)
    return (
        objective_value = objective_value(model),
        x1 = value(x[1]),
        x2 = value(x[2]),
        labor_shadowprice = shadow_price(labor),
        wood_shadowprice = shadow_price(wood),
    )
end
```

```julia
labor_avail = 30.0
wood_avail = 120.0
sol = solve_labor_wood_LP(labor_avail, wood_avail)
xmax = 30
ymax = 30
Plots.plot(
    xlabel="x_1 (Tables)", 
    ylabel="x_2 (Chairs)",  
    legend=:topright,
    aspect_ratio=:equal,
    xlims=(0, xmax),
    ylims=(0, ymax),
)
feasible_region = Shape(
    [ 
        (0, 0),
        (wood_avail / 10, 0),
        Tuple([1 2; 10 5] \ [labor_avail; wood_avail]),
        (0, labor_avail / 2),
        (0, 0)
    ]
)
Plots.plot!(
    feasible_region, 
    fillcolor=:gray,
    fillalpha=0.5,
    label=false,
)
Plots.plot!(
    [(x, (labor_avail-x)/2) for x in 0:0.1:xmax], 
    label="x + 2y = $(labor_avail)", 
    linewidth=3,
)
Plots.plot!(
    [(x, (wood_avail-10*x)/5) for x in 0:0.1:xmax], 
    label="10x + 5y = $(wood_avail)", 
    linewidth=3,
)
Plots.plot!(
    [(0, y) for y in 0:0.1:ymax], 
    label="x = 0", 
    linewidth=3,
)
Plots.plot!(
    [(x, 0) for x in 0:0.1:xmax], 
    label="y = 0", 
    linewidth=3,
)
Plots.scatter!(
    [sol.x1], 
    [sol.x2], 
    label="Optimal Solution", 
    color=:black, 
    markersize=5
)
Plots.annotate!(
    sol.x1 + 4, 
    sol.x2 + 1, 
    text(
        "($(string(round(sol.x1, digits=3))), $(string(round(sol.x2, digits=3))))",
        :black,
        8,
        :center,
    )
)
```
